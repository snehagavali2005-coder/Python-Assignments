```python
a = input("Enter a string: ")

longest = ""

for i in range(len(a)):
    for j in range(i + 1, len(a) + 1):
        sub = a[i:j]

        if sub == sub[::-1]:
            if len(sub) > len(longest):
                longest = sub

print("One longest palindromic substring =", longest)
```

```python
a = input("Enter encoded string: ")

result = ""

i = 0

while i < len(a):
    char = a[i]
    count = int(a[i + 1])

    result += char * count

    i += 2

print("Decoded string =", result)
```

```python
a = input("Enter an expression: ")

stack = []

for char in a:
    if char in "([{":
        stack.append(char)

    elif char in ")]}":
        if not stack:
            print("Balanced =", False)
            break

        top = stack.pop()

        if (char == ')' and top != '(') or \
           (char == ']' and top != '[') or \
           (char == '}' and top != '{'):
            print("Balanced =", False)
            break
else:
    print("Balanced =", len(stack) == 0)
```

```python
a = (1, 1, 2, 3, 3, 3)

result = ()

for num in a:
    found = False

    for item in result:
        if item[0] == num:
            found = True

    if not found:
        result = result + ((num, a.count(num)),)

print("Value and frequency =", result)
```

```python
a = ((1, 2, 3), (4, 5, 6))
result = ()

for i in range(len(a[0])):
    row = ()

    for j in range(len(a)):
        row = row + (a[j][i],)

    result = result + (row,)
print("Transpose =", result)
```

```python
a = ((1, 2, 3), (4, 5, 6))
result = ()

for i in range(len(a[0])):
    row = ()

    for j in range(len(a) - 1, -1, -1):
        row = row + (a[j][i],)

    result = result + (row,)
print("Rotated matrix =", result)
```

```python
a = (1, 2)
b = ('A', 'B')

result = ()
for x in a:
    for y in b:
        result = result + ((x, y),)

print("Cartesian product =", result)
```

```python
a = (1, 2, 3, 4)
result = ()

for i in range(len(a)):
    for j in range(i + 1, len(a)):
        result = result + ((a[i], a[j]),)

print("Combinations =", result)
```

```python
a = (1, (2, (3, 4)), 5)
result = ()

def flatten(t):
    global result

    for item in t:
        if isinstance(item, tuple):
            flatten(item)
        else:
            result = result + (item,)

flatten(a)
print("Flattened tuple =", result)
```

```python
a = (('Amit', 'DS'), ('Neha', 'AI'), ('Ravi', 'DS'))

result = {}

for name, branch in a:
    if branch not in result:
        result[branch] = ()
    result[branch] = result[branch] + (name,)

for branch in result:
    print(branch, "=", result[branch])
```

```python
a = (1, 2, 3, 4, 5)
n = 2

result = a[n:] + a[:n]
print("Rotated tuple =", result)
```

```python
a = (1, 2, 5, 3, 4, 6, 7, 2)

longest = (a[0],)
current = (a[0],)

for i in range(1, len(a)):
    if a[i] > a[i - 1]:
        current = current + (a[i],)
    else:
        if len(current) > len(longest):
            longest = current
        current = (a[i],)

if len(current) > len(longest):
    longest = current

print("Subsequence =", longest)
print("Length =", len(longest))
```

```python
a = (1, 4, 7)
b = (2, 3, 8, 9)

result = ()
i = 0
j = 0

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        result = result + (a[i],)
        i += 1
    else:
        result = result + (b[j],)
        j += 1

while i < len(a):
    result = result + (a[i],)
    i += 1

while j < len(b):
    result = result + (b[j],)
    j += 1

print("Merged tuple =", result)
```
